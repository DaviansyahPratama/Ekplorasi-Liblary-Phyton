## 2. Implementasi pada Permasalahan Probabilitas

### Studi Kasus 1: Teorema Bayes (Deteksi Malware pada Scanner Antivirus)

```python
# Studi Kasus 1: Teorema Bayes (Deteksi Malware pada Scanner Antivirus)
p_malware = 0.05           # P(M)    : 5% file yang diunggah adalah malware
p_aman = 1 - p_malware     # P(A)    : 95% file lainnya aman

p_deteksi_malware = 0.90   # P(D|M)  : scanner menandai 90% file malware
p_deteksi_aman = 0.04      # P(D|A)  : scanner salah menandai 4% file aman

# Probabilitas total sebuah file ditandai scanner (Hukum Probabilitas Total)
p_deteksi = (p_deteksi_malware * p_malware) + (p_deteksi_aman * p_aman)

# Teorema Bayes: P(M|D) = P(D|M) * P(M) / P(D)
p_malware_jika_deteksi = (p_deteksi_malware * p_malware) / p_deteksi

print(f"Probabilitas sebuah file ditandai scanner  : {p_deteksi:.4f}")
print(f"Probabilitas file benar-benar malware      : {p_malware_jika_deteksi:.4f}")
print(f"Dalam persen                               : {p_malware_jika_deteksi * 100:.2f}%")
```

### Studi Kasus 2: Kombinasi Kejadian (Pemilihan Test Case Acak)

```python
import math

# Studi Kasus 2: Kombinasi Kejadian (Pemilihan Test Case Acak)
total_test = 12                    # Jumlah seluruh test case
test_bug = 4                       # Test case yang akan mengungkap bug
test_normal = total_test - test_bug
dipilih = 3                        # Test case yang dijalankan secara acak

# Jumlah seluruh kemungkinan pemilihan (ruang sampel)
total_kombinasi = math.comb(total_test, dipilih)

# Kejadian A: tepat 2 test case mengungkap bug (2 bug dan 1 normal)
kombinasi_tepat_2 = math.comb(test_bug, 2) * math.comb(test_normal, 1)
prob_tepat_2 = kombinasi_tepat_2 / total_kombinasi

# Kejadian B: minimal 1 bug = 1 - P(tidak ada bug sama sekali)
prob_minimal_1 = 1 - math.comb(test_normal, dipilih) / total_kombinasi

print(f"Total kombinasi pemilihan   : {total_kombinasi}")
print(f"Kombinasi tepat 2 bug       : {kombinasi_tepat_2}")
print(f"P(tepat 2 bug)              : {prob_tepat_2:.4f} ({prob_tepat_2 * 100:.2f}%)")
print(f"P(minimal 1 bug)            : {prob_minimal_1:.4f} ({prob_minimal_1 * 100:.2f}%)")
```

### Studi Kasus 3: Simulasi Monte Carlo vs Perhitungan Matematis (Load Balancer)

```python
import random
from scipy.stats import binom

random.seed(42)
jumlah_simulasi = 10000
hari_aman = 0
prob_uptime = 0.85

# 1. Pendekatan Simulasi Acak (Monte Carlo)
for _ in range(jumlah_simulasi):
    # Setiap server bernilai True (aktif) jika angka acak (0.0 - 1.0) <= 0.85
    server_aktif = sum(random.random() <= prob_uptime for _ in range(5))

    # Layanan aman jika minimal 3 dari 5 server aktif
    if server_aktif >= 3:
        hari_aman += 1

prob_simulasi = hari_aman / jumlah_simulasi

# 2. Pendekatan Matematis Teoritis (Distribusi Binomial)
# Peluang aman = P(tepat 3 aktif) + P(tepat 4 aktif) + P(tepat 5 aktif)
prob_matematis = sum(binom.pmf(k, 5, prob_uptime) for k in (3, 4, 5))

print(f"Probabilitas layanan aman (Simulasi {jumlah_simulasi} iterasi): {prob_simulasi:.4f}")
print(f"Probabilitas layanan aman (Perhitungan Matematis): {prob_matematis:.4f}")
```

## 3. Implementasi pada Distribusi Probabilitas

### Studi Kasus 4: Distribusi Binomial (Access Point WiFi)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameter Distribusi Binomial
n_ap = 15            # Jumlah access point WiFi di gedung kampus
p_gangguan = 0.2     # Probabilitas sebuah AP terganggu dalam sehari
target_k = 4         # Target: tepat 4 AP mengalami gangguan

# Menghitung probabilitas spesifik (PMF) dan statistik distribusi
prob_tepat_4 = binom.pmf(target_k, n_ap, p_gangguan)
prob_min_8 = binom.sf(7, n_ap, p_gangguan)      # P(X >= 8)
mean_binom = binom.mean(n_ap, p_gangguan)
var_binom = binom.var(n_ap, p_gangguan)

print(f"Probabilitas tepat {target_k} AP terganggu : {prob_tepat_4:.4f} ({prob_tepat_4 * 100:.2f}%)")
print(f"Probabilitas minimal 8 AP terganggu : {prob_min_8:.4f} ({prob_min_8 * 100:.2f}%)")
print(f"Nilai Harapan (Ekspektasi)          : {mean_binom:.1f} AP")
print(f"Varians Distribusi                  : {var_binom:.1f}")

# Visualisasi Distribusi Binomial
x_values = np.arange(0, n_ap + 1)
pmf_values = binom.pmf(x_values, n_ap, p_gangguan)

plt.figure(figsize=(8, 5))
plt.bar(x_values, pmf_values, color="skyblue", edgecolor="black", alpha=0.8)
plt.bar(target_k, prob_tepat_4, color="orange", edgecolor="black", label=f"Tepat {target_k} AP")
plt.axvline(mean_binom, color="red", linestyle="--", label=f"Nilai Harapan = {mean_binom:.1f}")
plt.title("Distribusi Binomial: Gangguan Access Point WiFi")
plt.xlabel("Jumlah AP yang Terganggu")
plt.ylabel("Probabilitas (PMF)")
plt.xticks(x_values)
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

### Studi Kasus 5: Distribusi Normal (Waktu Loading Halaman Web)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameter Distribusi Normal
mean_load = 2.4       # Rata-rata waktu loading halaman (detik)
std_load = 0.5        # Standar deviasi (detik)
batas_tunggu = 3.2    # Batas kesabaran pengguna (detik)

# Probabilitas pengguna menunggu lebih dari 3.2 detik (area ekor kanan)
prob_lambat = norm.sf(batas_tunggu, loc=mean_load, scale=std_load)
z_score = (batas_tunggu - mean_load) / std_load
persentil_95 = norm.ppf(0.95, loc=mean_load, scale=std_load)

print(f"Z-score batas {batas_tunggu} detik            : {z_score:.2f}")
print(f"Probabilitas loading > {batas_tunggu} detik   : {prob_lambat:.4f} ({prob_lambat * 100:.2f}%)")
print(f"Persentil ke-95 waktu loading       : {persentil_95:.2f} detik")

# Visualisasi Distribusi Normal dan Area Kritis
x = np.linspace(mean_load - 4 * std_load, mean_load + 4 * std_load, 500)
y = norm.pdf(x, mean_load, std_load)

plt.figure(figsize=(8, 5))
plt.plot(x, y, color="purple", label="Kurva PDF Waktu Loading")
plt.fill_between(x, y, where=(x > batas_tunggu), color="red", alpha=0.5,
                 label=f"Area Kritis (> {batas_tunggu} detik)")
plt.axvline(mean_load, color="black", linestyle="--", label=f"Rata-rata = {mean_load} detik")
plt.title("Distribusi Normal: Waktu Loading Halaman Web")
plt.xlabel("Waktu Loading (detik)")
plt.ylabel("Kepadatan Probabilitas (Densitas)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
